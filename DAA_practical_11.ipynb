```python

INF = float('inf')

# Adjacency matrix representation of graph
graph = [
    [0,   5,   INF, 10],
    [INF, 0,   3,   INF],
    [INF, INF, 0,   1],
    [INF, INF, INF, 0]
]

V = len(graph)

# Floyd-Warshall algorithm
dist = [row[:] for row in graph]

for k in range(V):
    for i in range(V):
        for j in range(V):
            dist[i][j] = min(
                dist[i][j],
                dist[i][k] + dist[k][j]
            )

# Display shortest distance matrix
print("Shortest Distance Matrix:")
for row in dist:
    print(["INF" if x == INF else x for x in row])
```
