```python
# Kruskal's Algorithm

def find(parent, i):
    if parent[i] == i:
        return i
    return find(parent, parent[i])


def union(parent, x, y):
    parent[find(parent, x)] = find(parent, y)


# Number of vertices
V = 4

# Edges: (weight, vertex1, vertex2)
edges = [
    (10, 0, 1),
    (6, 0, 2),
    (5, 0, 3),
    (15, 1, 3),
    (4, 2, 3)
]

# Sort edges by weight
edges.sort()

parent = list(range(V))
mst = []
total_cost = 0

# Kruskal's Algorithm
for weight, u, v in edges:
    x = find(parent, u)
    y = find(parent, v)

    if x != y:
        mst.append((u, v, weight))
        total_cost += weight
        union(parent, x, y)

# Display MST
print("Edges in Minimum Spanning Tree:")

for u, v, weight in mst:
    print(u, "--", v, "=", weight)

print("Minimum Cost =", total_cost)
```
